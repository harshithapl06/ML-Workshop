```python
data.isnull().sum()
```

```python
data['Age']=data['Age'].fillna(data['Age'].mean(),inplace=True)
```

```python
data.isnull().sum()
```

```python
gender=pd.get_dummies(data['Sex'],drop_first=True)
gender
```

```python
pclass=pd.get_dummies(data['Pclass'],drop_first=True,dtype=int)
pclass
```

```python
embarked=pd.get_dummies(data['Embarked'],drop_first=True)
embarked
```

```python
final_data=pd.concat([data,pclass,gender,embarked],axis=1)
final_data
```

```python
final_data.columns
```

```python
final_data.drop(columns=['PassengerId','Name','Pclass','Sex','Ticket','Embarked'],inplace=True)
final_data.head()
```
